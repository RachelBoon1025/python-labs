## Question 3

```python
def heat_energy(mass, temperature_rise):
    energy = mass * 4180 * temperature_rise
    return energy
```

```python
mass = float(input("Enter the mass: "))
temperature_rise = float(input("Enter the temperature rise: "))

print("Energy", heat_energy(mass, temperature_rise), "J")
```
